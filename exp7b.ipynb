## PROGRAM

This experiment uses **Microsoft Power BI Desktop** (Windows). **No coding is needed** – every visual is made by **drag and drop**.

For each visual: (1) click an empty area of the page, (2) click the visual's icon in **Visualizations**, (3) drag the fields from the **Data** pane into the boxes under Visualizations.

| Visual | Icon in Visualizations | Drag field → into box |
|---|---|---|
| Card 1 – Total Revenue | Card | Sales → **Value** |
| Card 2 – Total Profit | Card | Profit → **Value** |
| Card 3 – Total Orders | Card | OrderID → **Value**, then click ⌄ → **Count (Distinct)** |
| Bar chart | Clustered bar chart | Category → **Y-axis**, Sales → **X-axis** |
| Line chart | Line chart | Date → Month → **X-axis**, Sales → **Y-axis** |
| Donut chart | Donut chart | Category → **Legend**, Profit → **Values** |
| Slicer 1 | Slicer | Region → **Field** |
| Slicer 2 | Slicer | Category → **Field** |

Number columns (marked **Σ**) are added up automatically, so Sales gives **Sum of Sales** = Total Revenue.

## STEP-BY-STEP WITH SCREENSHOTS

**A. Install Power BI Desktop (one time, Windows only)**

A1. Click **Start** – **Microsoft Store** (or type Microsoft Store in the search box).



A2. In the Store, type **Power BI Desktop**. Choose **Power BI Desktop** (not Power BI Report Builder).



A3. **Power BI Desktop** by Microsoft Corporation – click **Install** (or **Get**).



A4. It downloads about 900 MB. Wait until it finishes.



A5. When it shows **Open**, click **Open**.



_If the Store is blocked: open https://www.microsoft.com/en-us/download/details.aspx?id=58494 – **Download** – choose **PBIDesktopSetup_x64.exe** – run it – **Next**, accept the terms – **Install** – **Finish**._

**B. Load the sales data**

B1. On the start screen, click **Excel workbook**.



B2. Go to the lab folder – **data** – select **sales_data.xlsx** – **Open**.



B3. In the Navigator, tick **Sheet1** and click **Load**.



B4. The data is loaded. Click **>** next to Sheet1 in the Data pane to see the columns.



**C. Build each visual by drag and drop** – (1) click an empty area of the page, (2) click the visual's icon in Visualizations, (3) drag the fields into the boxes.

C1. **Card**: click the Card icon (5th row, 1st icon). An empty card and the **Value** box appear.



C2. Drag **Sales** into **Value**. The card shows Sum of Sales (Total Revenue). Make two more cards: **Profit** → Value, **OrderID** → Value.



C3. OrderID first shows **First OrderID**. In the Value box, click the small down arrow next to it and choose **Count (Distinct)**.



C4. Three cards: Total Revenue 5M, Total Profit 2M, Total Orders 200.



C5. **Bar chart**: click **Clustered bar chart** (1st row, 3rd icon).



C6. Drag **Category** into **Y-axis** and **Sales** into **X-axis**.



C7. **Line chart** (2nd row, 1st icon): Date → **Month** into **X-axis**, **Sales** into **Y-axis**.



C8. **Donut chart** (3rd row, last icon): **Category** into **Legend**, **Profit** into **Values**.



C9. **Slicer** (5th row, 3rd icon): drag **Region** into **Field**.



C10. Add a second **Slicer**: drag **Category** into **Field**.



C11. Save: **File – Save As – SuperMart_Sales_Dashboard**. If a slicer shows a tick you did not make, hover on it and click the **eraser** icon (Clear selections).

**D. Now do Task 1 and Task 2** (see the TASK QUESTION at the top). Take a screenshot with **Windows + Shift + S** for your record.
