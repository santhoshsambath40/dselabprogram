To create basic plots — Line, Bar, Scatter and Pie — using the Matplotlib library.

ALGORITHM
Step 1: Start the program.

Step 2: Import matplotlib.pyplot as plt and numpy as np.

Step 3: Enable inline plotting using %matplotlib inline.

Step 4: Create sample x and y data arrays.

Step 5: Draw a Line plot with title, x-label, y-label and grid.

Step 6: Draw a Bar chart with coloured bars.

PROGRAM
Click the code and press Shift + Enter to run it. The output appears just below the code.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 15, 7, 20, 12])

# 1. Line plot
plt.figure(figsize=(7, 4))
plt.plot(x, y, color='blue', marker='o', linewidth=2, label='y values')  # line plot of y against x
plt.title('Line Plot'); plt.xlabel('X'); plt.ylabel('Y')
plt.grid(True); plt.legend(); plt.show()

# 2. Bar chart
plt.figure(figsize=(7, 4))
plt.bar(x, y, color='green', edgecolor='black')               # bar chart
plt.title('Bar Chart'); plt.xlabel('X'); plt.ylabel('Y')
plt.show()

# 3. Scatter plot
plt.figure(figsize=(7, 4))
plt.scatter(x, y, color='red', s=100, marker='*')             # scatter plot; s = point size
plt.title('Scatter Plot'); plt.xlabel('X'); plt.ylabel('Y')
plt.grid(True); plt.show()

# 4. Pie chart
plt.figure(figsize=(6, 6))
labels = ['A', 'B', 'C', 'D', 'E']
plt.pie(y, labels=labels, autopct='%1.1f%%', startangle=90)   # autopct shows % on each slice
plt.title('Pie Chart'); plt.show()
```
