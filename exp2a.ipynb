```python
import os

# Create data folder
os.makedirs('data', exist_ok=True)

# Create a text file and write content
with open('data/myfile.txt', 'w') as file:
    file.write("Hello Python\n")
    file.write("Welcome to Programming\n")
    file.write("File Handling")
```

```python
# Read the whole file
with open('data/myfile.txt', 'r') as file1:
    print("Output of read():")
    print(file1.read())


# Read line by line
with open('data/myfile.txt', 'r') as file2:
    print("\nOutput of readline():")
    print(file2.readline())


# Read into a list
with open('data/myfile.txt', 'r') as file3:
    print("\nOutput of readlines():")
    print(file3.readlines())
```
